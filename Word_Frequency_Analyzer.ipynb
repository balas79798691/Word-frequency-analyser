# 🔤 Word Frequency Analyzer

## 🎯 Objective

Analyze a text and find the words that occur most frequently.


```python
import re
from collections import Counter

text = input("Enter your text: ")

if not text.strip():
    print("Please enter some text.")
else:
    # Extract words and convert them to lowercase
    words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

    # Common English words to ignore
    stop_words = {
        "the", "is", "a", "an", "and", "or", "to", "of",
        "in", "on", "for", "with", "this", "that", "are",
        "was", "were", "it", "as", "be", "by", "from"
    }

    filtered_words = [word for word in words if word not in stop_words]
    frequency = Counter(filtered_words)

    print("\n===== WORD FREQUENCY ANALYZER =====")
    print("Total words:", len(words))
    print("Unique words:", len(set(words)))
    print("\nMost Frequent Words:")

    for word, count in frequency.most_common(10):
        print(f"{word}: {count}")
```

## 💻 Example

**Input:** `Python is easy to learn. Python is powerful and Python is popular.`

The program counts each word and displays the most frequently used words.


## 🧠 Concepts Used

**Tokenization**, text preprocessing, stop-word removal, and frequency counting.


## ✅ Conclusion

The project analyzes text and identifies the words that appear most frequently.
